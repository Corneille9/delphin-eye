# Entraînement Delphin-Eye

Fine-tune le modèle `yolo26n.pt` sur les images de dauphins annotées (classe : `dorsal_fin`).
Les poids du meilleur modèle sont sauvegardés dans `output/runs/delphin-eye/weights/best.pt`.

```python
from ultralytics import YOLO
```

```python
from pathlib import Path

DATA_YAML = str(Path.cwd().parent / "data" / "data.yaml")
OUTPUT_RUNS = str(Path.cwd().parent / "output" / "models")

model = YOLO("yolo26n.pt")

model.train(
    data=DATA_YAML,
    epochs=50,
    imgsz=640,
    batch=16,
    project=OUTPUT_RUNS,
    name="default",
)
```
