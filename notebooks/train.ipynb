```python
from ultralytics import YOLO
```

```python
from pathlib import Path

DATA_YAML = str(Path.cwd().parent / "data" / "data.yaml")
OUTPUT = str(Path.cwd().parent / "output" / "models" / "default")

model = YOLO("../output/models/default/yolo26n.pt")

results = model.train(
    data=DATA_YAML,
    epochs=50,
    imgsz=640,
    batch=16,
    project=str(Path.cwd().parent / "output" / "runs"),
    name="delphin-eye",
)

metrics = model.val()
print(metrics)
```
