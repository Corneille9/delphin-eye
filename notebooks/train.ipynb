# Training - Delphin-Eye

Fine-tunes `yolo26s.pt` on annotated dolphin images (class: `dorsal_fin`).
Best weights are saved to `output/models/<name>/weights/best.pt`.

```python
from ultralytics import YOLO
from pathlib import Path

DATA_YAML = str(Path.cwd().parent / "data" / "data.yaml")
OUTPUT_RUNS = str(Path.cwd().parent / "output" / "models")

model = YOLO("yolo26s.pt")

model.train(
    data=DATA_YAML,
    epochs=200,
    patience=40,
    imgsz=960,
    batch=16,
    degrees=5,
    fliplr=0.5,
    flipud=0.0, # dorsal fins are never upside-down
    scale=0.5,
    translate=0.1,
    mosaic=1.0,
    close_mosaic=10,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    cache="disk",
    save_period=10,
    project=OUTPUT_RUNS,
    name="delv2-i725-e150-small",
)
```
